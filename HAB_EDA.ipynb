# Task 1 - HAB prediction: exploratory data analysis

The target is `HAB` (0 = no occurrence, 1 = occurrence). This notebook answers one question from section 3.1 of the
lab manual: **are the classes balanced?**

It also looks at how the imbalance varies by year, season and spatial cluster. Finally, it repeats the check on rows with
complete CHL values, because the manual says the 2021-2023 test set contains only complete samples. The last two analysis sections show the
distributions of `CHL` and `CHL_uncertainty`.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

DATA_PATH = Path("classification_task1_train.csv")
TARGET = "HAB"
CHL_COLUMNS = ["CHL", "CHL_uncertainty"]
GROUPS = {"Year": "Year", "Season": "Season", "spatial_cluster": "Spatial cluster"}
SEASON_ORDER = ["Winter", "Spring", "Summer", "Fall"]

BAR_COLOR = "#2a78d6"
REFERENCE_COLOR = "#52514e"
GRID_COLOR = "#e4e3df"

plt.rcParams.update({
    "figure.dpi": 110,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.edgecolor": GRID_COLOR,
    "axes.axisbelow": True,
    "font.size": 10,
})
```

## 1. Load the data

```python
df = pd.read_csv(DATA_PATH, encoding="utf-8-sig", parse_dates=["Date"])
df = df.sort_values("Date", kind="stable").reset_index(drop=True)

print(f"{df.shape[0]} rows, {df.shape[1]} columns, {df['Date'].min():%Y-%m-%d} to {df['Date'].max():%Y-%m-%d}")
print(f"Missing target values: {df[TARGET].isna().sum()}")
df.head()
```

## 2. Overall class balance

```python
def class_balance(data):
    """Return the count and share of each HAB class."""
    counts = data[TARGET].value_counts().sort_index()
    return pd.DataFrame({"count": counts, "share (%)": (counts / counts.sum() * 100).round(1)})


overall = class_balance(df)
negatives, positives = overall["count"]

print(f"Imbalance ratio (class 0 : class 1): {negatives / positives:.2f} : 1")
print(f"Accuracy of always predicting class 0: {negatives / len(df):.1%}")
overall
```

```python
fig, ax = plt.subplots(figsize=(7, 2.2))
labels = ["HAB = 0 (no occurrence)", "HAB = 1 (occurrence)"]
ax.barh(labels, overall["count"], color=BAR_COLOR, height=0.55)
ax.invert_yaxis()
for i, (count, share) in enumerate(zip(overall["count"], overall["share (%)"])):
    ax.text(count + 10, i, f"{count} ({share}%)", va="center")
ax.set_xlim(0, overall["count"].max() * 1.2)
ax.set_xlabel("Number of samples")
ax.xaxis.grid(True, color=GRID_COLOR)
ax.set_title(f"Class 0 outnumbers class 1 by {negatives / positives:.1f} to 1", loc="left")
fig.tight_layout()
plt.show()
```

## 3. HAB rate by year, season and spatial cluster

A single overall number can hide large differences between groups. Each bar below shows the percentage of samples with
`HAB = 1`, labelled as positives / samples. Small groups give unreliable rates, so read the rate together with its label.

```python
def hab_summary(data, column):
    """Return samples, positives and HAB rate for each value of a column."""
    summary = data.groupby(column)[TARGET].agg(samples="size", positives="sum")
    summary["rate"] = (summary["positives"] / summary["samples"] * 100).round(1)
    return summary.reindex(SEASON_ORDER) if column == "Season" else summary


def plot_hab_rate(data, title):
    """Draw one panel of HAB rates per grouping, on a shared y-axis."""
    overall_rate = data[TARGET].mean() * 100
    fig, axes = plt.subplots(1, 3, figsize=(12, 3.8), sharey=True, width_ratios=[8, 4, 6])
    for ax, (column, label) in zip(axes, GROUPS.items()):
        summary = hab_summary(data, column)
        positions = list(range(len(summary)))
        ax.bar(positions, summary["rate"], color=BAR_COLOR, width=0.65)
        ax.axhline(overall_rate, color=REFERENCE_COLOR, linestyle="--", linewidth=1)
        for i, row in enumerate(summary.itertuples()):
            ax.text(i, row.rate + 1, f"{row.positives}/{row.samples}", ha="center", fontsize=8, zorder=5,
                    bbox={"facecolor": "white", "edgecolor": "none", "pad": 1})
        ax.set_xticks(positions, summary.index.astype(str))
        ax.set_xlabel(label)
        ax.yaxis.grid(True, color=GRID_COLOR)
        ax.tick_params(length=0)
    axes[0].set_ylabel("Samples with HAB = 1 (%)")
    axes[0].set_ylim(0, 55)
    axes[-1].text(1, overall_rate + 1, f"overall {overall_rate:.1f}%", ha="center",
                  color=REFERENCE_COLOR, fontsize=8)
    fig.suptitle(title, x=0.01, ha="left")
    fig.tight_layout()
    plt.show()


plot_hab_rate(df, "HAB rate varies strongly by year, season and cluster (all rows)")
```

```python
for column in GROUPS:
    display(hab_summary(df, column).rename(columns={"rate": "HAB rate (%)"}))
```

## 4. Rows with complete CHL values

The test set contains only complete samples, so the complete rows here are the closest match to it in the training data.

```python
is_complete = df[CHL_COLUMNS].notna().all(axis=1)
complete = df[is_complete]

comparison = pd.DataFrame({
    "All rows": [len(df), df[TARGET].sum()],
    "Complete CHL": [len(complete), complete[TARGET].sum()],
    "Missing CHL": [(~is_complete).sum(), df.loc[~is_complete, TARGET].sum()],
}, index=["samples", "positives"]).T
comparison["HAB rate (%)"] = (comparison["positives"] / comparison["samples"] * 100).round(1)
comparison["ratio (0 : 1)"] = ((comparison["samples"] - comparison["positives"]) / comparison["positives"]).round(2)

print(f"Rows missing CHL: {(~is_complete).mean():.1%}")
comparison
```

```python
plot_hab_rate(complete, "The same pattern holds on rows with complete CHL values")
```

```python
for column in GROUPS:
    display(hab_summary(complete, column).rename(columns={"rate": "HAB rate (%)"}))
```

## 5. Distribution of CHL values

This section uses only the rows where `CHL` is present. The table summarises the values for all of those rows and for each
HAB class, and the plots then show the shape of the distribution on a linear and on a log scale.

```python
chl = complete["CHL"]
chl_groups = {
    "All rows": chl,
    "HAB = 0": chl[complete[TARGET] == 0],
    "HAB = 1": chl[complete[TARGET] == 1],
}


def upper_outliers(values):
    """Count values above the upper Tukey fence (Q3 + 1.5 * IQR)."""
    q1, q3 = values.quantile([0.25, 0.75])
    return (values > q3 + 1.5 * (q3 - q1)).sum()


chl_summary = pd.DataFrame({
    name: values.describe(percentiles=[0.05, 0.25, 0.5, 0.75, 0.95]) for name, values in chl_groups.items()
}).T
chl_summary["skewness"] = [values.skew() for values in chl_groups.values()]
chl_summary["skewness (log10)"] = [np.log10(values).skew() for values in chl_groups.values()]

print(f"Non-positive CHL values: {(chl <= 0).sum()}")
print(f"Values above the upper Tukey fence: {upper_outliers(chl)} on the linear scale, "
      f"{upper_outliers(np.log10(chl))} on the log10 scale")
chl_summary.round(2)
```

```python
CHL_BINS = 30
LOG_BINS = np.geomspace(chl.min(), chl.max(), CHL_BINS + 1)
LOG_TICKS = [0.5, 1, 2, 5, 10, 20, 50]


def format_log_axis(ax):
    """Use a log x-axis with plain numbers as tick labels."""
    ax.set_xscale("log")
    ax.set_xticks(LOG_TICKS, [str(tick) for tick in LOG_TICKS])
    ax.minorticks_off()


def mark_median(ax, values):
    """Draw and label the median of the values on a histogram."""
    median = values.median()
    ax.axvline(median, color=REFERENCE_COLOR, linestyle="--", linewidth=1)
    ax.text(median * 1.04, ax.get_ylim()[1] * 0.92, f"median {median:.2f}", color=REFERENCE_COLOR, fontsize=8)


fig, (ax_linear, ax_log) = plt.subplots(1, 2, figsize=(12, 3.8))

ax_linear.hist(chl, bins=CHL_BINS, color=BAR_COLOR, edgecolor="white", linewidth=0.8)
ax_linear.set_xlabel("CHL")
ax_linear.set_title("Linear scale", loc="left", fontsize=10)

ax_log.hist(chl, bins=LOG_BINS, color=BAR_COLOR, edgecolor="white", linewidth=0.8)
format_log_axis(ax_log)
ax_log.set_xlabel("CHL (log scale)")
ax_log.set_title("Log scale", loc="left", fontsize=10)

ax_linear.set_ylabel("Number of samples")
for ax in (ax_linear, ax_log):
    ax.yaxis.grid(True, color=GRID_COLOR)
    ax.tick_params(length=0)
    mark_median(ax, chl)

fig.suptitle(f"CHL is strongly right-skewed, and much less so on a log scale ({len(chl)} samples)", x=0.01, ha="left")
fig.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True, sharey=True)

for ax, name in zip(axes, ["HAB = 0", "HAB = 1"]):
    values = chl_groups[name]
    ax.hist(values, bins=LOG_BINS, weights=np.full(len(values), 100 / len(values)),
            color=BAR_COLOR, edgecolor="white", linewidth=0.8)
    ax.set_ylabel("Share of class (%)")
    ax.set_title(f"{name} ({len(values)} samples)", loc="left", fontsize=10)
    ax.yaxis.grid(True, color=GRID_COLOR)
    ax.tick_params(length=0)

format_log_axis(axes[-1])
axes[-1].set_xlabel("CHL (log scale)")
for ax, name in zip(axes, ["HAB = 0", "HAB = 1"]):
    mark_median(ax, chl_groups[name])

fig.suptitle("CHL tends to be higher when HAB = 1, but the two classes overlap heavily", x=0.01, ha="left")
fig.tight_layout()
plt.show()
```

## 6. Distribution of CHL_uncertainty

This repeats the checks from section 5 on `CHL_uncertainty`, to see whether it is skewed like `CHL`. It uses the same rows,
because the two columns are always missing together.

```python
uncertainty = complete["CHL_uncertainty"]
uncertainty_groups = {
    "All rows": uncertainty,
    "HAB = 0": uncertainty[complete[TARGET] == 0],
    "HAB = 1": uncertainty[complete[TARGET] == 1],
}

uncertainty_summary = pd.DataFrame({
    name: values.describe(percentiles=[0.05, 0.25, 0.5, 0.75, 0.95]) for name, values in uncertainty_groups.items()
}).T
uncertainty_summary["skewness"] = [values.skew() for values in uncertainty_groups.values()]
uncertainty_summary["skewness (log10)"] = [np.log10(values).skew() for values in uncertainty_groups.values()]

print(f"Non-positive values: {(uncertainty <= 0).sum()}")
print(f"Values above the upper Tukey fence: {upper_outliers(uncertainty)} on the linear scale, "
      f"{upper_outliers(np.log10(uncertainty))} on the log10 scale")
uncertainty_summary.round(2)
```

```python
fig, (ax_linear, ax_log) = plt.subplots(1, 2, figsize=(12, 3.8))

ax_linear.hist(uncertainty, bins=CHL_BINS, color=BAR_COLOR, edgecolor="white", linewidth=0.8)
ax_linear.set_xlabel("CHL_uncertainty")
ax_linear.set_title("Linear scale", loc="left", fontsize=10)

uncertainty_ticks = [10, 20, 50, 90]
ax_log.hist(uncertainty, bins=np.geomspace(uncertainty.min(), uncertainty.max(), CHL_BINS + 1),
            color=BAR_COLOR, edgecolor="white", linewidth=0.8)
ax_log.set_xscale("log")
ax_log.set_xticks(uncertainty_ticks, [str(tick) for tick in uncertainty_ticks])
ax_log.minorticks_off()
ax_log.set_xlabel("CHL_uncertainty (log scale)")
ax_log.set_title("Log scale", loc="left", fontsize=10)

ax_linear.set_ylabel("Number of samples")
for ax in (ax_linear, ax_log):
    ax.yaxis.grid(True, color=GRID_COLOR)
    ax.tick_params(length=0)
    mark_median(ax, uncertainty)

fig.suptitle(f"CHL_uncertainty is only mildly right-skewed and has two peaks ({len(uncertainty)} samples)", x=0.01, ha="left")
fig.tight_layout()
plt.show()
```

## 7. Findings

**The classes are imbalanced.**
- There are 925 samples of class 0 and 138 of class 1, so only 13.0% are positive (a ratio of 6.70 : 1).
- Always predicting class 0 reaches 87.0% accuracy while finding no HAB events, so accuracy alone is not a useful measure here.
  On the manual's primary metric, AUC, a model that gives every sample the same score gets only 0.5.
- Rows with complete CHL values are slightly less imbalanced: 14.7% positive (5.78 : 1), against 11.0% in rows missing CHL.

**The imbalance isn't uniform.** These figures are for all rows:
- **Year:** the HAB rate ranges from 0% to 47.4%. 2014 and 2017 have no positives, and 2015 has 27 in 57 samples.
- **Season:** the rate goes from 2.9% in winter to 27.8% in summer.
- **Spatial cluster:** cluster 2 has the highest rate (35.4%), and clusters 1 and 3 have no positives but only 13 samples between them.

**CHL is strongly right-skewed.** These figures are for the 563 rows with a CHL value:
- **Range:** values go from 0.35 to 62.73. The median is 2.93, the mean is 5.66 and the 95th percentile is 20.68.
- **Shape:** skewness falls from 3.80 to 0.49 on a log10 scale. All values are positive, so the log is defined for every row.
- **High values:** 57 values are above the upper Tukey fence on the linear scale, but only 9 on the log scale, so most of
  them are part of the long tail rather than clear outliers.
- **By class:** the median is 4.25 for `HAB = 1` and 2.73 for `HAB = 0`, but the middle 50% of values overlap (2.55-9.59
  and 1.74-5.56). The `HAB = 1` histogram has only 83 samples, so its finer detail is noisy.

**`CHL_uncertainty` is much less skewed than `CHL`.** These figures are for the same 563 rows:
- **Shape:** skewness is 0.53, against 3.80 for `CHL`, and -0.04 on a log10 scale. The histogram has two peaks, one at about 15-25 and one at about 65-75.
- **Range:** values go from 6.74 to 92.74, with a median of 32.48 and a mean of 39.03. No values are above the upper Tukey fence.
- **By class:** the median is 40.46 for `HAB = 1` and 31.36 for `HAB = 0`, and the middle 50% of values overlap (25.80-67.86 and 19.35-56.02).

**What this means for the next steps:**
- Validation folds need checking for positives, because a fold holding only 2014 or 2017 has no positives and its AUC is undefined.
- A chronological split will see a different class balance in each fold, so a single split could give a misleading score.
- Class weighting or resampling is worth comparing during model selection, applied to the training folds only.
- A log transform of CHL is worth trying for models sensitive to feature scale, and the Tukey fence on the linear scale
  shouldn't be used on its own to remove high CHL values.
- `CHL_uncertainty` doesn't need a log transform for skew. Its two peaks won't be removed by a log transform or by z-scoring.
- CHL shifts between the classes but doesn't separate them, so it's unlikely to be enough as a single predictor.

**Differences from the manual:**
- The day-of-year column is called `YDOY` in the data, not `DOY`.
- CHL is missing in 47.0% of rows, while the manual says 46%.
